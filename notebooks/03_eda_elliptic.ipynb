## Per time step

Transaction count and illicit share for each of the 49 time steps. The two measures have different scales, so they get separate stacked panels sharing the x-axis rather than a dual axis. Illicit share is illicit / (illicit + licit): unknown nodes are excluded, since their class is not known.

```python
import matplotlib.pyplot as plt

by_step = nodes.groupby('time_step').agg(
    txns=('txId', 'size'),
    illicit=('label', lambda s: (s == 1).sum()),
    licit=('label', lambda s: (s == 0).sum()),
)
by_step['labeled'] = by_step['illicit'] + by_step['licit']
by_step['illicit_share'] = by_step['illicit'] / by_step['labeled'] * 100
assert by_step.index.min() == 1 and by_step.index.max() == 49 and len(by_step) == 49

INK, MUTED, GRID = '#1f2328', '#656d76', '#e6e8eb'
BAR, LINE = '#4f7cac', '#c2553a'

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 6.5), sharex=True, gridspec_kw={'hspace': 0.18})
ax1.bar(by_step.index, by_step['txns'], color=BAR, width=0.7)
ax1.set_title('Transactions per time step', loc='left', color=INK, fontsize=11)
ax1.set_ylabel('Transactions', color=MUTED)
ax1.yaxis.set_major_formatter(lambda v, _: f'{v:,.0f}')

ax2.plot(by_step.index, by_step['illicit_share'], color=LINE, lw=2, marker='o', ms=4, mec='white', mew=1)
ax2.set_title('Illicit share of labeled transactions', loc='left', color=INK, fontsize=11)
ax2.set_ylabel('Illicit %', color=MUTED)
ax2.set_ylim(bottom=0)
ax2.set_xlabel('Time step', color=MUTED)
ax2.set_xticks(range(1, 50, 2))

for ax in (ax1, ax2):
    ax.grid(axis='y', color=GRID, lw=0.8)
    ax.set_axisbelow(True)
    ax.tick_params(colors=MUTED, length=0)
    for side in ('top', 'right', 'left'):
        ax.spines[side].set_visible(False)
    ax.spines['bottom'].set_color(GRID)
plt.show()
```

## Dark market shutdown (time step 43)

A dark market was shut down around step 43, after which the illicit share is expected to collapse. Check whether the drop comes from fewer illicit nodes, a change in volume, or a change in labeling coverage.

```python
SHUTDOWN = 43

step_labels = by_step.assign(unknown=by_step['txns'] - by_step['labeled'])
step_labels['unknown_share'] = step_labels['unknown'] / step_labels['txns'] * 100
step_labels.loc[38:49, ['txns', 'illicit', 'licit', 'unknown_share', 'illicit_share']]
```

```python
period = np.where(by_step.index < SHUTDOWN, f'before (1-{SHUTDOWN - 1})', f'after ({SHUTDOWN}-49)')
cmp = by_step.groupby(period).agg(
    steps=('txns', 'size'), txns=('txns', 'sum'), illicit=('illicit', 'sum'), licit=('licit', 'sum'),
)
cmp['txns_per_step'] = cmp['txns'] / cmp['steps']
cmp['illicit_per_step'] = cmp['illicit'] / cmp['steps']
cmp['unknown_share'] = (1 - (cmp['illicit'] + cmp['licit']) / cmp['txns']) * 100
cmp['illicit_share'] = cmp['illicit'] / (cmp['illicit'] + cmp['licit']) * 100
cmp
```

## Feature groups

Elliptic ships 166 features per transaction: the time step, 93 local features (transaction-only) and 72 aggregated features (one-hop neighbour statistics). The time step is stored in `time_step` here, so `f0`–`f164` hold 165 features: `f0`–`f92` are local and `f93`–`f164` are aggregated. Counting `time_step` as a local feature gives the 94 local features quoted in the paper.

```python
N_LOCAL, N_AGG = 93, 72
LOCAL_FEATURES = FEATURES[:N_LOCAL]
AGG_FEATURES = FEATURES[N_LOCAL:]
assert len(LOCAL_FEATURES) + len(AGG_FEATURES) == len(FEATURES) == 165
assert len(AGG_FEATURES) == N_AGG and not set(LOCAL_FEATURES) & set(AGG_FEATURES)

print(f'local:      {len(LOCAL_FEATURES)} ({LOCAL_FEATURES[0]}..{LOCAL_FEATURES[-1]}), {len(LOCAL_FEATURES) + 1} with time_step')
print(f'aggregated: {len(AGG_FEATURES)} ({AGG_FEATURES[0]}..{AGG_FEATURES[-1]})')
```

## Illicit vs licit feature distributions

On labeled nodes only, rank each feature by how differently it is distributed between the two classes. Two measures: Cohen's d (standardized mean difference, signed: positive = higher for illicit) and the two-sample Kolmogorov-Smirnov statistic (largest gap between the class CDFs, 0-1, catches shape differences that means miss). Ranking is by KS.

```python
from scipy.stats import ks_2samp

lab = nodes[nodes['label'].isin([0, 1])]
ill, lic = lab[lab['label'] == 1], lab[lab['label'] == 0]

def cohens_d(a, b):
    pooled = np.sqrt(((len(a) - 1) * a.var() + (len(b) - 1) * b.var()) / (len(a) + len(b) - 2))
    return (a.mean() - b.mean()) / pooled

diff = pd.DataFrame({
    'group': ['local' if f in set(LOCAL_FEATURES) else 'aggregated' for f in FEATURES],
    'illicit_median': ill[FEATURES].median(),
    'licit_median': lic[FEATURES].median(),
    'cohens_d': [cohens_d(ill[f], lic[f]) for f in FEATURES],
    'ks': [ks_2samp(ill[f], lic[f]).statistic for f in FEATURES],
}, index=FEATURES)
diff['abs_d'] = diff['cohens_d'].abs()
diff = diff.sort_values('ks', ascending=False)
print(f'{len(ill):,} illicit vs {len(lic):,} licit')
diff.head(20).drop(columns='abs_d')
```

```python
diff.groupby('group')[['ks', 'abs_d']].agg(['mean', 'max']).round(3)
```

```python
top = diff.head(6).index
fig, axes = plt.subplots(2, 3, figsize=(11, 6), sharey=False)
for ax, f in zip(axes.flat, top):
    lo, hi = lab[f].quantile([0.01, 0.99])
    bins = np.linspace(lo, hi, 40)
    ax.hist(lic[f].clip(lo, hi), bins=bins, density=True, color=BAR, alpha=0.6, label='licit')
    ax.hist(ill[f].clip(lo, hi), bins=bins, density=True, color=LINE, alpha=0.6, label='illicit')
    ax.set_title(f'{f} ({diff.loc[f, "group"]}, KS={diff.loc[f, "ks"]:.2f})', loc='left', color=INK, fontsize=10)
    ax.tick_params(colors=MUTED, length=0)
    ax.set_yticks([])
    for side in ('top', 'right', 'left'):
        ax.spines[side].set_visible(False)
    ax.spines['bottom'].set_color(GRID)
axes[0, 0].legend(frameon=False)
fig.suptitle('Top features by KS: illicit vs licit (density, clipped to 1st-99th percentile)', x=0.01, ha='left', color=INK, fontsize=11)
fig.tight_layout()
plt.show()
```

## Graph basics

Edges are directed (`txId1` -> `txId2`, money flowing between transactions). Build a sparse adjacency matrix over node indices, then look at size, degree distribution and connected components.

```python
from scipy.sparse import coo_matrix
from scipy.sparse.csgraph import connected_components

idx = pd.Series(np.arange(len(nodes)), index=nodes['txId'])
src_i, dst_i = idx[edges['txId1']].to_numpy(), idx[edges['txId2']].to_numpy()
N = len(nodes)
A = coo_matrix((np.ones(len(edges), dtype=np.int8), (src_i, dst_i)), shape=(N, N)).tocsr()

deg = pd.DataFrame({'in': np.asarray(A.sum(axis=0)).ravel(), 'out': np.asarray(A.sum(axis=1)).ravel()}, index=nodes['txId'])
deg['total'] = deg['in'] + deg['out']

print(f'nodes: {N:,}   edges: {len(edges):,}   density: {len(edges) / (N * (N - 1)):.2e}')
print(f'mean degree (in = out): {len(edges) / N:.2f}')
print(f'edges crossing time steps: {(nodes["time_step"].to_numpy()[src_i] != nodes["time_step"].to_numpy()[dst_i]).sum()}')
deg.describe(percentiles=[.5, .9, .99]).T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
for ax, col in zip(axes, ['in', 'out']):
    vc = deg[col].value_counts().sort_index()
    ax.loglog(np.maximum(vc.index, 0.5), vc.values, 'o', color=BAR, ms=4, mec='white', mew=0.5)
    ax.set_title(f'{col}-degree distribution (log-log)', loc='left', color=INK, fontsize=11)
    ax.set_xlabel('degree (0 shown at 0.5)', color=MUTED)
    ax.set_ylabel('nodes', color=MUTED)
    ax.grid(color=GRID, lw=0.8)
    ax.set_axisbelow(True)
    ax.tick_params(colors=MUTED, length=0)
    for side in ('top', 'right'):
        ax.spines[side].set_visible(False)
fig.tight_layout()
plt.show()
print('top 10 total degree:'); deg.sort_values('total', ascending=False).head(10)
```

```python
n_wcc, wcc = connected_components(A, directed=False)
n_scc, scc = connected_components(A, directed=True, connection='strong')
sizes = pd.Series(wcc).value_counts()

print(f'weakly connected components: {n_wcc:,}')
print(f'strongly connected components: {n_scc:,} (= nodes: {n_scc == N}, i.e. the graph is acyclic)' if n_scc == N else f'strongly connected components: {n_scc:,}')
print(f'largest WCC: {sizes.iloc[0]:,} nodes ({sizes.iloc[0] / N:.1%})')
print(f'singleton WCCs: {(sizes == 1).sum():,}')

comp = pd.DataFrame({'wcc': wcc, 'time_step': nodes['time_step'].to_numpy(), 'label': nodes['label'].to_numpy()})
print(f'WCCs spanning >1 time step: {(comp.groupby("wcc")["time_step"].nunique() > 1).sum()}')
sizes.describe(percentiles=[.5, .9, .99]).to_frame('WCC size').T
```

```python
fig, ax = plt.subplots(figsize=(6, 3.8))
vc = sizes.value_counts().sort_index()
ax.loglog(vc.index, vc.values, 'o', color=BAR, ms=4, mec='white', mew=0.5)
ax.set_title('Weakly connected component sizes (log-log)', loc='left', color=INK, fontsize=11)
ax.set_xlabel('component size (nodes)', color=MUTED)
ax.set_ylabel('components', color=MUTED)
ax.grid(color=GRID, lw=0.8)
ax.set_axisbelow(True)
ax.tick_params(colors=MUTED, length=0)
for side in ('top', 'right'):
    ax.spines[side].set_visible(False)
plt.show()

top_wcc = comp[comp['wcc'].isin(sizes.index[:5])].groupby('wcc').agg(
    nodes=('label', 'size'), time_step=('time_step', 'first'),
    illicit=('label', lambda s: (s == 1).sum()), licit=('label', lambda s: (s == 0).sum()),
).sort_values('nodes', ascending=False)
top_wcc
```

## Neighbour label mix

For each edge, tabulate the label of the source and destination. Compare against what random mixing would give (the same label frequencies, edges ignored): if illicit nodes link to illicit ones more often than chance, the graph is homophilous and neighbour information is useful.

```python
lbl = nodes['label'].map(LABEL_NAMES).to_numpy()
pairs = pd.DataFrame({'src': lbl[src_i], 'dst': lbl[dst_i]})
order = ['illicit', 'licit', 'unknown']
mix = pd.crosstab(pairs['src'], pairs['dst']).reindex(index=order, columns=order)
print('edge counts (rows = source label, columns = destination label)')
display(mix)
print('row-normalised: where do edges from each class go (%)')
display((mix.div(mix.sum(axis=1), axis=0) * 100).round(1))
```

```python
share = nodes['label'].map(LABEL_NAMES).value_counts(normalize=True).reindex(order) * 100
mix_pct = mix.div(mix.sum(axis=1), axis=0) * 100
lift = mix_pct.div(share, axis=1)
print('lift = observed share of destinations / overall share of that class (1.0 = random mixing)')
display(lift.round(2))

# labeled-only view: of edges between two labeled nodes, how many are illicit-illicit etc.
both = mix.loc[['illicit', 'licit'], ['illicit', 'licit']]
ill_share = (nodes['label'] == 1).sum() / nodes['label'].isin([0, 1]).sum() * 100
print(f'labeled-to-labeled edges only, row-normalised (%); overall illicit share of labeled = {ill_share:.1f}%')
display((both.div(both.sum(axis=1), axis=0) * 100).round(1))
```

```python
# per-node view: for illicit and licit nodes, fraction of (in+out) neighbours that are illicit / licit / unknown
nbr = pd.concat([
    pd.DataFrame({'node': src_i, 'nbr': lbl[dst_i]}),
    pd.DataFrame({'node': dst_i, 'nbr': lbl[src_i]}),
])
nbr_counts = pd.crosstab(nbr['node'], nbr['nbr']).reindex(index=range(N), columns=order, fill_value=0)
nbr_counts['own'] = lbl
per_node = nbr_counts.assign(deg=nbr_counts[order].sum(axis=1))
for c in order:
    per_node[f'frac_{c}'] = per_node[c] / per_node['deg']
per_node[per_node['own'].isin(['illicit', 'licit'])].groupby('own')[[f'frac_{c}' for c in order] + ['deg']].mean().round(3)
```

## Time-based split

Train on time steps 1-34, test on 35-49, labeled nodes only (unknown nodes are dropped from both sets, but stay in the graph for neighbour features). Splitting by time avoids leaking future transactions into training; since no edge crosses time steps, no graph information leaks either.

```python
TRAIN_STEPS, TEST_STEPS = range(1, 35), range(35, 50)

is_labeled = nodes['label'].isin([0, 1])
train_mask = is_labeled & nodes['time_step'].isin(TRAIN_STEPS)
test_mask = is_labeled & nodes['time_step'].isin(TEST_STEPS)
assert not (train_mask & test_mask).any()
assert nodes.loc[train_mask, 'time_step'].max() < nodes.loc[test_mask, 'time_step'].min()

split = pd.DataFrame({
    name: {
        'steps': f'{m.start}-{m.stop - 1}',
        'labeled nodes': int(mask.sum()),
        'illicit': int((nodes.loc[mask, 'label'] == 1).sum()),
        'licit': int((nodes.loc[mask, 'label'] == 0).sum()),
        'illicit share %': (nodes.loc[mask, 'label'] == 1).mean() * 100,
        'unknown dropped': int(((nodes['time_step'].isin(m)) & ~is_labeled).sum()),
    } for name, m, mask in [('train', TRAIN_STEPS, train_mask), ('test', TEST_STEPS, test_mask)]
}).T
split['share of labeled %'] = split['labeled nodes'].astype(int) / is_labeled.sum() * 100
split
```

```python
test_by_step = by_step.loc[list(TEST_STEPS), ['labeled', 'illicit', 'illicit_share']]
post = test_by_step.loc[SHUTDOWN:]
print(f'test illicit in steps 35-{SHUTDOWN - 1}: {test_by_step.loc[:SHUTDOWN - 1, "illicit"].sum():,}')
print(f'test illicit in steps {SHUTDOWN}-49 (after shutdown): {post["illicit"].sum():,}')
test_by_step.T
```

### Save split

One row per labeled node (`txId`, `time_step`, `label`, `split`). Features are not duplicated here; join to `elliptic_nodes.parquet` on `txId`.

```python
OUT = ROOT / 'data' / 'processed' / 'elliptic_split.parquet'
OUT.parent.mkdir(parents=True, exist_ok=True)

split_df = nodes.loc[train_mask | test_mask, ['txId', 'time_step', 'label']].copy()
split_df['split'] = np.where(train_mask[split_df.index], 'train', 'test')
split_df['label'] = split_df['label'].astype('int8')
split_df.to_parquet(OUT, index=False)

chk = pd.read_parquet(OUT)
assert len(chk) == train_mask.sum() + test_mask.sum() and chk['txId'].is_unique
print(f'wrote {OUT.relative_to(ROOT)}: {len(chk):,} rows')
chk.groupby('split')['label'].agg(['size', 'sum'])
```

## Key findings

1. **Clean data.** 203,769 transactions and 234,355 directed edges, with no nulls, duplicate `txId`s, duplicate edges, self-loops or edges to unknown nodes. There are 165 features (`f0`-`f164`) plus `time_step`: 93 local (`f0`-`f92`) and 72 aggregated (`f93`-`f164`).
2. **Few labels, heavy imbalance.** 77.2% of nodes are unknown, 20.6% licit and only 2.2% illicit (4,545). Among labeled nodes, 9.8% are illicit.
3. **The illicit share varies a lot over time.** It ranges from 0.3% to 36% per time step, so a single global rate is misleading.
4. **The dark market shutdown (step 43) cuts illicit labels.** Illicit nodes per step fall from about 104 (steps 1-42) to about 24 (steps 43-49), and the illicit share of labeled nodes from 11.0% to 2.5%. Volume and the unknown share are unchanged, so this is fewer illicit labels, not less data. Steps 45-46 have only 5 and 2 illicit nodes.
5. **Local features separate the classes better than aggregated ones.** Mean KS is 0.34 for local and 0.23 for aggregated; the top 12 features by KS are all local (best: f52, KS 0.62). Many top features are near-duplicates, and several differ in shape rather than mean (high KS, tiny Cohen's d).
6. **Each time step is its own graph.** There are exactly 49 weakly connected components, one per time step, with no edges between steps and no directed cycles. Degrees are heavy-tailed (max out-degree 472, max in-degree 284). A time-based split cannot leak through edges.
7. **The graph is homophilous.** Illicit nodes link to illicit nodes 13x more often than chance, but about half of their labeled links go to licit nodes (52% illicit, 48% licit). Licit nodes almost never link to illicit ones (2.3% of their labeled links), so an illicit neighbour is a strong signal.
8. **The time split is shifted and the test set is dominated by pre-shutdown steps.** Train (steps 1-34) has 29,894 labeled nodes at 11.6% illicit; test (35-49) has 16,670 at 6.5%. Of the 1,083 test illicit nodes, 914 are in steps 35-42 and 169 in steps 43-49. Report metrics overall and separately for these two ranges.
